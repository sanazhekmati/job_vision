```python
import pandas as pd
job_vision = pd.read_csv(
    r"F:\sanazi\python\data science\code\JobVision_Jobposts_Dataset\JobVision_Jobposts_Dataset.csv"
)
job_vision
```

```python
import pandas as pd
import matplotlib.pyplot as plt


job_vision["AverageSalary"] = (
    job_vision["Jobpost_MinSalary"] +
    job_vision["Jobpost_MaxSalary"]
) / 2


job_vision["HasLanguageSkill"] = (
    job_vision["Jobpost_LanguageSkills"]
    .notna()
)


language_salary = (
    job_vision
    .groupby("HasLanguageSkill")
    .agg(
        Number_of_Posts=("AverageSalary", "count"),
        Average_Salary=("AverageSalary", "mean")
    )
)

print(language_salary)


plt.figure(figsize=(6,5))

language_salary["Average_Salary"].plot(kind="bar")

plt.title("Average Salary by Language Skill Requirement")
plt.xlabel("Language Skill Required")
plt.ylabel("Average Salary")
plt.xticks([0,1], ["No", "Yes"], rotation=0)

plt.show()
```
