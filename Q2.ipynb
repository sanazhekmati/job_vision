```python
import pandas as pd


df = job_vision[["Jobpost_IsRemote", "Jobpost_MinSalary", "Jobpost_MaxSalary"]].copy()


df = df.dropna()

df["AverageSalary"] = (df["Jobpost_MinSalary"] + df["Jobpost_MaxSalary"]) / 2


result = (
    df.groupby("Jobpost_IsRemote")["AverageSalary"]
      .mean()
)

print(result)
```

```python
import matplotlib.pyplot as plt

df = job_vision[["Jobpost_IsRemote", "Jobpost_MinSalary", "Jobpost_MaxSalary"]].dropna()
df["AverageSalary"] = (df["Jobpost_MinSalary"] + df["Jobpost_MaxSalary"]) / 2

df.boxplot(column="AverageSalary", by="Jobpost_IsRemote")

plt.title("Salary Distribution")
plt.suptitle("")
plt.xlabel("Remote")
plt.ylabel("Average Salary")
plt.show()
```
