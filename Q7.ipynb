```python
import pandas as pd
job_vision = pd.read_csv(
    r"F:\sanazi\python\data science\code\JobVision_Jobposts_Dataset\JobVision_Jobposts_Dataset.csv"
)
job_vision
```

```python
print(job_vision["Jobpost_SoftwareSkills"].dropna().head(10))
```

```python
skills = (
    job_vision["Jobpost_SoftwareSkills"]
    .dropna()
    .str.split(",")
    .explode()
    .str.strip()
)

top_skills = skills.value_counts().reset_index()
top_skills.columns = ["Skill", "Count"]

print(top_skills.head(10))
```

```python
import matplotlib.pyplot as plt

top_skills.head(10).plot(
    x="Skill",
    y="Count",
    kind="bar",
    figsize=(10,5),
    legend=False
)

plt.title("Top 10 Most Demanded Software Skills")
plt.xlabel("Software Skill")
plt.ylabel("Number of Job Posts")
plt.xticks(rotation=45)
plt.show()
```
