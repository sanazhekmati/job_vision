```python
import pandas as pd
job_vision = pd.read_csv(
    r"F:\sanazi\python\data science\code\JobVision_Jobposts_Dataset\JobVision_Jobposts_Dataset.csv"
)
job_vision
```

```python
job_vision["AverageSalary"] = (
    job_vision["Jobpost_MinSalary"] +
    job_vision["Jobpost_MaxSalary"]
) / 2
```

```python
size_salary = (
    job_vision
    .groupby("Company_SizeEn")
    .agg(
        Number_of_Posts=("AverageSalary", "count"),
        Average_Salary=("AverageSalary", "mean")
    )
    .sort_values("Average_Salary", ascending=False)
)

print(size_salary)
```

```python
import matplotlib.pyplot as plt

size_salary["Average_Salary"].plot(kind="bar", figsize=(8,5))

plt.title("Average Salary by Company Size")
plt.xlabel("Company Size")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)
plt.show()
```
